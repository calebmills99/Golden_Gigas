## 📦 What IS npm?

npm (Node Package Manager) is the world's largest software registry. It allows developers to:

- **Share** code packages with other developers
- **Reuse** code from other developers
- **Manage** project dependencies
- **Version control** packages

Think of it like Sophia's Sicily story: instead of building everything from scratch, you can use pre-built packages that others have created and tested!

## 📥 Installing Packages

There are several ways to install packages with npm:

````python
# Installing packages - Command examples

examples = {
    "Install a package locally": "npm install express",
    "Install as a dev dependency": "npm install --save-dev jest",
    "Install globally": "npm install -g create-react-app",
    "Install specific version": "npm install lodash@4.17.21",
    "Install all dependencies from package.json": "npm install"
}

print("📦 npm Installation Commands:\n")
for description, command in examples.items():
    print(f"▶ {description}:")
    print(f"  $ {command}\n")
````

## 📊 Dependencies vs. DevDependencies

**SOPHIA:** "Let me break it down for you, Dorothy."

- **Dependencies**: Packages your app needs to RUN (like flour for making bread)
- **DevDependencies**: Packages you only need for DEVELOPMENT (like the recipe book - you don't serve it with the bread!)

````python
# Example package.json with dependencies

package_with_deps = {
    "name": "golden-girls-retirement-website",
    "version": "1.0.0",
    "dependencies": {
        "express": "^4.18.2",          # Web framework - needed to run
        "react": "^18.2.0",            # UI library - needed to run
        "axios": "^1.4.0"              # HTTP client - needed to run
    },
    "devDependencies": {
        "jest": "^29.5.0",             # Testing framework - only for development
        "eslint": "^8.42.0",           # Linter - only for development
        "webpack": "^5.88.0"           # Bundler - only for development
    }
}

print("📄 Package.json with Dependencies:\n")
print(json.dumps(package_with_deps, indent=2))
````

## ⚡ npm Scripts

npm scripts are custom commands you can define in your package.json file:

````python
# Example package.json with useful scripts

package_with_scripts = {
    "name": "golden-girls-retirement-website",
    "version": "1.0.0",
    "scripts": {
        "start": "node server.js",                          # Start the application
        "dev": "nodemon server.js",                        # Start with auto-reload
        "test": "jest",                                    # Run tests
        "test:watch": "jest --watch",                      # Run tests in watch mode
        "build": "webpack --mode production",              # Build for production
        "lint": "eslint .",                                # Check code quality
        "lint:fix": "eslint . --fix",                      # Fix code quality issues
        "clean": "rm -rf dist node_modules",               # Clean build artifacts
        "deploy": "npm run build && npm run upload"        # Chain multiple commands
    }
}

print("⚡ Package.json with Scripts:\n")
print(json.dumps(package_with_scripts, indent=2))
print("\n🏃 To run a script, use: npm run <script-name>")
print("   Example: npm run dev")
````

## 🔍 Understanding Version Numbers (Semantic Versioning)

**SOPHIA:** "Pay attention, Dorothy. This is important."

Version numbers in npm follow the format: **MAJOR.MINOR.PATCH**

- **MAJOR** (1.0.0): Breaking changes - like changing the recipe entirely
- **MINOR** (0.1.0): New features - like adding chocolate chips to cookies
- **PATCH** (0.0.1): Bug fixes - like fixing a typo in the recipe

**Special Symbols:**
- **^1.2.3** - Compatible with 1.2.3, allows MINOR and PATCH updates (1.x.x)
- **~1.2.3** - Approximately 1.2.3, allows only PATCH updates (1.2.x)
- **1.2.3** - Exactly version 1.2.3

````python
# Semantic versioning examples

versioning_examples = {
    "^1.2.3": "Allows: 1.2.3, 1.2.4, 1.3.0, 1.9.9 | Blocks: 2.0.0",
    "~1.2.3": "Allows: 1.2.3, 1.2.4, 1.2.9 | Blocks: 1.3.0, 2.0.0",
    "1.2.3": "Allows: Only 1.2.3 exactly",
    "*": "Allows: Any version (not recommended!)",
    ">=1.2.3": "Allows: 1.2.3 and any higher version",
    "1.2.x": "Allows: 1.2.0, 1.2.1, etc. | Blocks: 1.3.0"
}

print("📊 Semantic Versioning Guide:\n")
for version, description in versioning_examples.items():
    print(f"▶ {version:12} → {description}")
````

## 📚 Quick Reference Guide

### Essential npm Commands (Dorothy's Cheat Sheet)

| Command | Description | When to Use |
|---------|-------------|-------------|
| `npm init` | Create a new project | Starting a new project |
| `npm init -y` | Create with defaults | Quick project setup |
| `npm install <package>` | Install a package | Adding functionality |
| `npm install` | Install all dependencies | After cloning a project |
| `npm uninstall <package>` | Remove a package | Cleaning up |
| `npm update` | Update packages | Keeping things current |
| `npm run <script>` | Run a script | Executing custom commands |
| `npm list` | Show installed packages | Checking what's installed |
| `npm outdated` | Check for updates | Finding old packages |
| `npm audit` | Security check | Finding vulnerabilities |
| `npm audit fix` | Fix security issues | Patching vulnerabilities |

````python
# Practical Example: Setting up a real project

print("🏗️ Setting Up a Real Project - Step by Step:\n")
print("1️⃣ Create project directory:")
print("   $ mkdir my-awesome-project")
print("   $ cd my-awesome-project")
print()
print("2️⃣ Initialize npm:")
print("   $ npm init -y")
print()
print("3️⃣ Install dependencies:")
print("   $ npm install express")
print("   $ npm install --save-dev nodemon")
print()
print("4️⃣ Create your main file:")
print("   $ touch index.js")
print()
print("5️⃣ Add scripts to package.json:")
print('   "scripts": {')
print('     "start": "node index.js",')
print('     "dev": "nodemon index.js"')
print('   }')
print()
print("6️⃣ Start developing:")
print("   $ npm run dev")
print()
print("✅ You're ready to code!")
````

## 🎓 Key Takeaways (What Dorothy Learned)

1. **npm is a package manager** - It helps you use and manage code libraries
2. **package.json is your project's blueprint** - It lists all dependencies and settings
3. **Dependencies are needed for production** - Your app needs these to run
4. **DevDependencies are for development only** - Testing, linting, building tools
5. **npm scripts automate common tasks** - Like shortcuts for complex commands
6. **Semantic versioning matters** - Understand ^, ~, and exact versions
7. **Always run `npm install`** - After cloning or pulling a project
8. **Use `npm audit`** - To check for security vulnerabilities
9. **Read the documentation** - Every package has docs on npmjs.com
10. **Don't commit node_modules** - Add it to .gitignore!
